
# Customer Segmentation and Personality Analysis

## Project Objective
The objective of this project is to analyze customer behavior and segment customers into meaningful groups using clustering techniques. 
This helps businesses design targeted marketing strategies and improve campaign effectiveness.



## 1. Import Required Libraries


```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

plt.show()
```


## 2. Load Dataset


```python

df = pd.read_excel("marketing_campaign1 (1).xlsx")
df.head()
```


## 3. Basic Data Understanding


```python

df.shape, df.info()
```

```python
df.columns
```


## 4. Data Cleaning
- Handle missing values
- Remove unnecessary records


```python

# Handling missing income values
df['Income'] = df['Income'].fillna(df['Income'].median())

# Remove duplicates if any
df = df.drop_duplicates()
```


## 5. Feature Engineering


```python

# Create Age feature
df['Age'] = 2025 - df['Year_Birth']

# Total spending feature
spend_cols = ['MntWines','MntFruits','MntMeatProducts','MntFishProducts','MntSweetProducts','MntGoldProds']
df['Total_Spending'] = df[spend_cols].sum(axis=1)

df[['Age','Total_Spending']].head()
```


## 6. Exploratory Data Analysis (EDA)


```python

plt.figure()
sns.histplot(df['Income'], kde=True)
plt.title("Income Distribution")
plt.show()
```

```python

plt.figure()
sns.scatterplot(x='Income', y='Total_Spending', data=df)
plt.title("Income vs Total Spending")
plt.show()
```

```python

plt.figure()
sns.countplot(x='Education', data=df)
plt.title("Education Distribution")
plt.xticks(rotation=45)
plt.show()
```


## 7. Feature Selection for Clustering


```python

cluster_features = [
    'Income','Age','Recency','NumWebPurchases',
    'NumStorePurchases','NumCatalogPurchases','Total_Spending'
]

X = df[cluster_features]
```


## 8. Feature Scaling


```python

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```


## 9. Finding Optimal Number of Clusters (Elbow Method)


```python

inertia = []
k_range = range(2, 11)

for k in k_range:
    km = KMeans(n_clusters=k, random_state=42)
    km.fit(X_scaled)
    inertia.append(km.inertia_)

plt.figure()
plt.plot(k_range, inertia, marker='o')
plt.xlabel("Number of Clusters")
plt.ylabel("Inertia")
plt.title("Elbow Method")
plt.show()
```

### Silhouette Score Analysis

The silhouette score is used to evaluate how well the data points are grouped into clusters.
A higher silhouette score indicates better-defined and well-separated clusters.
We calculate silhouette scores for different values of K to support the selection
of the optimal number of clusters.

```python
from sklearn.metrics import silhouette_score

for k in range(2, 8):
    kmeans = KMeans(n_clusters=k, random_state=42)
    labels = kmeans.fit_predict(X_scaled)
    score = silhouette_score(X_scaled, labels)
    print(f"K = {k}, Silhouette Score = {score:.3f}")
```

### Final K-Means Model

Based on the elbow method and silhouette score analysis, we selected 4 clusters
as the optimal number. The final K-Means model is trained using K = 4 and
cluster labels are assigned to each customer.



## 10. Apply K-Means Clustering


```python

kmeans = KMeans(n_clusters=4, random_state=42)
df['Cluster'] = kmeans.fit_predict(X_scaled)

df['Cluster'].value_counts()
```


## 11. Cluster Evaluation


```python

silhouette_score(X_scaled, df['Cluster'])
```

```python
from sklearn.metrics import davies_bouldin_score, calinski_harabasz_score

db_score = davies_bouldin_score(X_scaled, df['Cluster'])
ch_score = calinski_harabasz_score(X_scaled, df['Cluster'])

db_score, ch_score
```

```python
numeric_cols = df.select_dtypes(include=['int64', 'float64']).columns

cluster_profile = df.groupby('Cluster')[numeric_cols].mean()
cluster_profile
```

```python
df['Total_Purchases'] = (
    df['NumWebPurchases'] +
    df['NumStorePurchases'] +
    df['NumCatalogPurchases']
)
```

```python
df['Total_Spending'] = (
    df['MntWines'] +
    df['MntFruits'] +
    df['MntMeatProducts'] +
    df['MntFishProducts'] +
    df['MntSweetProducts'] +
    df['MntGoldProds']
)
```

```python
df['Deal_purchase_ratio'] = df['NumDealsPurchases'] / df['Total_Purchases']
```

```python
df['Deal_purchase_ratio'] = df['Deal_purchase_ratio'].fillna(0)
```

```python
cluster_profile = df.groupby('Cluster')[[
    'Income',
    'Recency',
    'Total_Spending',
    'Total_Purchases',
    'Deal_purchase_ratio'
]].mean()

cluster_profile
```


## 12. Cluster Analysis


```python

cluster_summary = df.groupby('Cluster')[cluster_features + ['Total_Spending']].mean()
cluster_summary
```


## 13. Visualization of Clusters


```python

plt.figure()
sns.scatterplot(x='Income', y='Total_Spending', hue='Cluster', data=df, palette='Set1')
plt.title("Customer Segments")
plt.show()
```

Interpretation of the Identified Clusters

After applying the KMeans clustering algorithm, the dataset was segmented into four distinct clusters based on standardized feature values such as Income and Total_Spending. Each cluster represents a unique customer behavior pattern, making the segmentation meaningful and interpretable.

Cluster 0: High Income – High Spending Customers

Customers in this cluster have high average income and high total spending.

Interpretation:
This cluster represents the most valuable customers. They have strong purchasing power and actively spend, contributing significantly to overall revenue.

Business Insight:
These customers are ideal targets for premium products, loyalty programs, and personalized high-value offers.

Cluster 1: High Income – Low Spending Customers

Customers in this cluster have high income but comparatively low spending.

Interpretation:
Although these customers earn well, they do not spend proportionally. This indicates cautious or selective spending behavior.

Business Insight:
This group has high potential. Targeted marketing strategies, personalized recommendations, and promotional offers can help convert them into high-spending customers.

Cluster 2: Low Income – High Spending Customers

Customers in this cluster have lower income levels but exhibit relatively high spending behavior.

Interpretation:
These customers tend to spend aggressively despite limited income, possibly driven by impulse buying or strong engagement with products.

Business Insight:
They are active buyers but may be price-sensitive. Budget-friendly offers, discounts, and value-based marketing strategies are suitable for this segment.

Cluster 3: Low Income – Low Spending Customers

Customers in this cluster have low income and low spending.

Interpretation:
This group contributes the least to revenue and shows limited purchasing activity.

Business Insight:
They are not a priority for high-cost marketing campaigns. Low-cost engagement strategies and awareness programs may be applied selectively.

Overall Conclusion

The four clusters represent all major combinations of income and spending behavior. This segmentation is well-defined, interpretable, and actionable, making it suitable for business decision-making and customer targeting strategies. Selecting four clusters provides an optimal balance between clustering performance and real-world interpretability.
